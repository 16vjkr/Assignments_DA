```python
#Question 1: Write a Python program that stores two numbers and
"""prints:
● Sum
● Difference
● Product
● Quotient
● Floor Division
● Modulus
● Exponent"""

try:
    #Declaring and accepting two float values from the user:
    number1 = float(input("Enter your first number: "))
    number2 = float(input("Enter your second number: "))
    #Displaying the arithmetic operations
    print(f"Sum: {number1 + number2}")
    print(f"Difference: {number1 - number2}")
    print(f"Product: {number1 * number2}")

#Handling zero as input
    if number2 !=0:
        print(f"Quotient: {number1 / number2}")
        print(f"Floor Division: {number1 // number2}")
        print(f"Modulus: {number1 % number2}")
    else: 
        print("Division operations cannot be performed with zero.")

    print(f"Exponent: {number1 ** number2}")
except ValueError:
    print("Please enter valid number.")
```

```python
#Question 2: Create a list of five fruits.
"""Perform the following:
● Add Mango
● Remove Banana
● Sort the list
● Reverse the list"""

#Creating a list of 5 fruits:
fruits = ["Apple", "Banana", "Cherry", "Dragon Fruit", "Emu Apple"]
print(f"Below are the list of 05 fruits: \n{fruits}")
#Adding Mango to the list:
fruits.append("Mango")
print(f"\nAdded Mango to the list of 05 fruits: \n{fruits}")
#Removing Banana from the list:
fruits.remove("Banana")
print(f"\nRemoved Banana from the list of 05 fruits: \n{fruits}")

#Reversing the list of fruits:
fruits.reverse()
print(f"\nReversing the list of fruits: {fruits}")

#Sorting the list
fruits.sort()
print(f"\nSorting the list in ascending order: {fruits}")

#Sorting the list again in descending order
fruits.sort(reverse=True)
print(f"\nSorting the list in descending order: {fruits}")
```

```python
#Question 9: Given data=[2,3,4,2,5,2,3,6,5] Create a dictionary containing the frequency of each element.
# Given list

data = [2, 3, 4, 2, 5, 2, 3, 6, 5]
# Creating an empty dictionary
frequency = {}
# Counting frequency of each element
for item in data:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
# Displaying the frequency dictionary
print("Frequency of each element:")
print(frequency)
```

```python
#Question 10: Write a function that accepts marks of five subjects
"""and returns
● Total
● Percentage
● Grade """

# Function to calculate total, percentage, and grade
def calculate_result(mark1, mark2, mark3, mark4, mark5):

    total = mark1 + mark2 + mark3 + mark4 + mark5
    percentage = total / 5

    if percentage >= 90:
        grade = "Grade A+"
    elif percentage >= 80:
        grade = "Grade A"
    elif percentage >= 70:
        grade = "Grade B"
    elif percentage >= 60:
        grade = "Grade C"
    elif percentage >= 50:
        grade = "Grade D"
    else:
        grade = "Grade F"
    return total, percentage, grade
try:
    # Accepting marks from the user
    mark1 = float(input("Enter marks of English: "))
    mark2 = float(input("Enter marks of Mathematics: "))
    mark3 = float(input("Enter marks of Computer: "))
    mark4 = float(input("Enter marks of Science: "))
    mark5 = float(input("Enter marks of Hindi: "))

    # Calling the function
    total, percentage, grade = calculate_result(
        mark1, mark2, mark3, mark4, mark5
    )
    # Displaying the result
    print(f"\nTotal Marks: {total}")
    print(f"Percentage: {percentage}%")
    print(f"Grade: {grade}")

except ValueError:
    print("Please enter valid marks.")
```

```python
#Question 13 : Generate a random 6-digit OTP.
"""Ask the user to enter the OTP.
Display whether verification is successful."""

# Importing random module
import random

# Generating a 6-digit OTP
otp = random.randint(100000, 999999)

print(f"Generated OTP: {otp}")

try:
    # Accepting OTP from the user
    user_otp = int(input("Enter the OTP: "))

    # Verifying OTP
    if user_otp == otp:
        print("OTP Verification Successful.")
    else:
        print("Invalid OTP. Verification Failed.")

except ValueError:
    print("Please enter a valid 6-digit OTP.")
```

```python
#Question 14 : Create reusable functions
"""● Deposit
● Withdraw
● Check Balance
Handle insufficient balance using exception handling."""

# Initial account balance
balance = 10000

# Function to deposit money
def deposit(balance, amount):
    balance += amount
    print(f"Amount Deposited: {amount}")
    return balance
# Function to withdraw money
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient Balance.")
    balance -= amount
    print(f"Amount Withdrawn: {amount}")
    return balance
# Function to check balance
def check_balance(balance):
    print(f"Available Balance: {balance}")
try:
    check_balance(balance)
    balance = deposit(balance, 5000)
    check_balance(balance)
    balance = withdraw(balance, 12000)
    check_balance(balance)
except ValueError as error:
    print(error)
```
